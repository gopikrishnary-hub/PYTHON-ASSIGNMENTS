```python
#1.question(finding average)
student_name = input("Enter the student name:")
student_id = int(input("Enter the student id:"))
Age = int(input("Enter the age:"))
Marks_1 = int(input("Enter the marks of first subject:"))
Marks_2 = int(input("Enter the marks of second subject:"))
Marks_3 = int(input("Enter the marks of third subject:"))

Total_marks = Marks_1 + Marks_2 + Marks_3
avg = Total_marks/3

if avg >=90:
    grade = "grade:A"
elif avg  >= 75:
    grade = "grade:B"
elif avg >= 60:
    grade = "grade:C"
elif avg >= 50:
    grade = "grade:D"
else:
    grade = "Fail"

print(f"Student Name:{student_name}")
print(f"Total Marks:{Total_marks}")
print(f"Average: {avg}")
print(f"{grade}")
```

```python
#question-5(Food Ordering System Using Match-Case )
print("welcome to gopi's restaurent..!!!!")

print("1. Pizza      - Rs. 250\n2. Burger     - Rs. 120\n3. Sandwich   - Rs. 100\n4. Fried Rice - Rs. 150\n5. Exit")

options = int(input("Enter your choice (1 to 5): "))

match options:
    case 1:
        print("You selected: Pizza - Rs. 250")
    case 2:
        print("You selected: Burger - Rs. 120")
    case 3:
        print("You selected: Sandwich - Rs. 100")
    case 4:
        print("You selected: Fried Rice - Rs. 150")
    case 5:
        print("Thank You! Visit Again")
    case _:
        print("Invalid Food Selection")
```
